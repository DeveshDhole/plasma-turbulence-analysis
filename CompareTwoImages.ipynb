```python
from PIL import Image
import matplotlib.pyplot as plt

# Load images
img1 = Image.open("SingleCurrentSheetAnalysis/J2_1D_Cut_NoFilter.png")
img2 = Image.open("SingleCurrentSheetAnalysis/J2_1D_Cut_Filter.png")

# Create a figure with 2 subplots
fig, axes = plt.subplots(1, 2, figsize=(10, 5))

# Show first image
axes[0].imshow(img1)
axes[0].axis("off")
axes[0].set_title("No Filter")

# Show second image
axes[1].imshow(img2)
axes[1].axis("off")
axes[1].set_title("Filter")

plt.savefig("IterativeFilteringComp", dpi=300)
plt.tight_layout()
plt.show()
```
