```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
i = j = np.arange(10)
```

```python
I, J = np.meshgrid(i, j)

n = I**2 + J**2

print(n)
```

```python
M1 = 2*np.sqrt(n)/16
```

```python
print(M1)
```

```python
M1 = np.sort(M1.flatten())
```

```python
print(M1)
```

```python
plt.scatter(M1, np.zeros_like(M1), s=5)
plt.xlabel(r'$n = i^2 + j^2$')
plt.yticks([])
plt.show()
```
